```r
library(datasets)
data(airquality)
airquality <- na.omit(airquality)

# Prueba de correlación de Pearson con nivel de significación del 5%
pearson_test_5 <- cor.test(airquality$Wind, airquality$Solar.R, method = "pearson", conf.level = 0.95)
p_value_5 <- pearson_test_5$p.value

# Prueba de correlación de Pearson con nivel de significación del 10%
pearson_test_10 <- cor.test(airquality$Wind, airquality$Solar.R, method = "pearson", conf.level = 0.90)
p_value_10 <- pearson_test_10$p.value

# Mostrar p-valores
p_value_5
p_value_10
```

El p-valor (0.1834) es mayor que 0.05, por lo que no rechazamos la hipótesis nula.
Esto significa que no hay evidencia suficiente para afirmar que existe una correlación lineal significativa entre Wind y Solar.R.
Con un nivel de significación del 10% (α = 0.10):

El p-valor (0.1834) también es mayor que 0.10, por lo que tampoco rechazamos la hipótesis nula.
Nuevamente, no encontramos evidencia de correlación significativa entre Wind y Solar.R.
No hay diferencias en la interpretación al cambiar el nivel de significación del 5% al 10%, ya que en ambos casos el p-valor sigue siendo mayor que el umbral crítico.
Por lo tanto, no hay suficiente evidencia estadística para concluir que existe una relación lineal significativa entre la velocidad del viento (Wind) y la radiación solar (Solar.R) en el dataset airquality.
Si queremos evaluar otro tipo de relación, podríamos considerar coeficientes no paramétricos como Spearman o Kendall.

<div style="background-color:#5DADE2"><b>Apartado 3</b></div>

```r
shapiro.test(airquality$Wind)  # Test de normalidad para Wind
shapiro.test(airquality$Solar.R)  # Test de normalidad para Solar.R

# Coeficientes de correlación de Spearman y Kendall con ambos niveles de significación
spearman_test_5 <- cor.test(airquality$Wind, airquality$Solar.R, method = "spearman", conf.level = 0.95)
spearman_test_10 <- cor.test(airquality$Wind, airquality$Solar.R, method = "spearman", conf.level = 0.90)

kendall_test_5 <- cor.test(airquality$Wind, airquality$Solar.R, method = "kendall", conf.level = 0.95)
kendall_test_10 <- cor.test(airquality$Wind, airquality$Solar.R, method = "kendall", conf.level = 0.90)

# Mostrar resultados
list(pearson_5 = pearson_test_5, pearson_10 = pearson_test_10,
     spearman_5 = spearman_test_5, spearman_10 = spearman_test_10,
     kendall_5 = kendall_test_5, kendall_10 = kendall_test_10)
```

Se realiza la prueba de Shapiro-Wilk para evaluar la normalidad de cada variable.
Si las variables no siguen una distribución normal, Pearson puede no ser adecuado.
Se calculan los coeficientes de correlación de Spearman y Kendall como alternativas.
Se comparan los resultados con los obtenidos con Pearson, analizando las diferencias.

Test de normalidad:

Si p-valor < 0.05, la variable no sigue una distribución normal.
Si p-valor > 0.05, la variable sigue una distribución normal.
Comparación de coeficientes de correlación:

Pearson es apropiado solo si las variables siguen una distribución normal.
Spearman y Kendall son más robustos para distribuciones no normales o datos ordinales.
Si los resultados de Spearman y Kendall son diferentes a Pearson, puede indicar que la relación entre las variables no es estrictamente lineal.

Generar una muestra aleatoria $x$ de $100$ valores de una distribución normal con $μ=3$ y $σ=0.4$ fijando antes `set.seed(2024)`, y una muestra aleatoria $y$ de $100$ valores de una distribución exponencial de parámetro $λ=0.9$ fijando antes `set.seed(2025)`. Se os pide:

* Contrastar si $x$ e $y$ provienen de una misma distribución. Tenéis que obtener el p-valor del test y justificar si, con un nivel de significación $α=0.05$, tendríamos que rechazar la hipótesis nula de que estas dos muestras provienen de la misma distribución.

* Realizar el mismo ejercicio pero simulando únicamente $10$ valores tanto para $x$ como para $y$ y realizar el mismo contraste. ¿Existe alguna diferencia en los resutlados obtenidos? ¿Qué podéis deducir de los resultados obtenidos en ambos casos?

<div style="background-color:#F7DC6F"><b>Resolución Ejercicio 04</b></div>

<div style="background-color:#5DADE2"><b>Apartado 1</b></div>

```r
set.seed(2024)
x <- rnorm(100, mean = 3, sd = 0.4)  # Muestra de distribución normal

set.seed(2025)
y <- rexp(100, rate = 0.9)  # Muestra de distribución exponencial

# Mostrar las primeras observaciones de cada muestra
head(x)
head(y)
summary(x)
summary(y)
```

Se generan dos muestras aleatorias de tamaño 100.
x sigue una distribución normal con media 3 y desviación estándar 0.4.
y sigue una distribución exponencial con parámetro lambda = 0.9.
Se fija la semilla para garantizar replicabilidad.

El p-valor = 1.515689e-32 es mucho menor que el nivel de significación α = 0.05.

Como p < 0.05, rechazamos la hipótesis nula (H₀). Esto significa que las muestras NO provienen de la misma distribución.

Esto era esperable porque:

x sigue una distribución normal con media 3 y desviación estándar 0.4.
y sigue una distribución exponencial, que es asimétrica y sesgada a la derecha.

<div style="background-color:#5DADE2"><b>Apartado 2</b></div>

```r
ks_test <- ks.test(x, y)  # Test de Kolmogorov-Smirnov
ks_test

# Mostrar resultado del test
print(ks_test)
```

Se usa el test de Kolmogorov-Smirnov para comparar si ambas muestras provienen de la misma distribución.
# El p-valor obtenido se compara con el nivel de significación α = 0.05.
Si p < 0.05, rechazamos la hipótesis nula y concluimos que las distribuciones son diferentes.

Hipótesis del contraste
Hipótesis nula (H₀): Las dos muestras provienen de la misma distribución.
Hipótesis alternativa (H₁): Las dos muestras provienen de distribuciones diferentes.
Nivel de significación (α)
En este caso, tomamos α = 0.05, que es el umbral comúnmente utilizado en estadística para decidir si un resultado es significativo.
Estadístico de prueba (D)
El estadístico D = 0.86 mide la mayor diferencia absoluta entre las funciones de distribución empíricas de ambas muestras.
Valores altos de D indican una gran diferencia entre las distribuciones.
P-valor (< 2.2e-16)
El p-valor representa la probabilidad de obtener un resultado igual o más extremo bajo la hipótesis nula.
Dado que el p-valor es prácticamente cero (mucho menor que 0.05), hay evidencia estadísticamente significativa para rechazar H₀.
Conclusión
Como el p-valor < 0.05, rechazamos H₀, lo que significa que las muestras provienen de distribuciones diferentes.
Este resultado tiene sentido, ya que x proviene de una distribución normal, mientras que y proviene de una distribución exponencial, que es inherentemente diferente.

<div style="background-color:#5DADE2"><b>Apartado 3</b></div>

```r
test_result <- ifelse(ks_test$p.value < 0.05, "Se rechaza H0: las muestras provienen de distribuciones diferentes", "No se rechaza H0: no hay evidencia suficiente para decir que las muestras son diferentes")
print(test_result)
```

Hipótesis nula (H0): Las dos muestras provienen de la misma distribución.

Nivel de significación: α = 0.05.
Estadístico de prueba: Distancia máxima entre las funciones de distribución empíricas de ambas muestras.
Decisión: Si el p-valor es menor que 0.05, rechazamos H0.
Al ser menor que 0.05, se rechaza H0.

3. ¿Qué par de variables tienen mayor correlación? ¿Qué par de variables tiene menor correlación? Realizar los contrastes de hipótesis de correlación oportunos para obtener las evidencias contrastadas dos a dos de dichos pares de variables. Justificar la respuesta proporcionada.

<div style="background-color:#F7DC6F"><b>Resolución Ejercicio 05</b></div>

<div style="background-color:#5DADE2"><b>Apartado 1</b></div>

```r
# Cargar los datos desde el archivo CSV (asumiendo que el archivo está en el mismo directorio)
calificaciones <- read.csv("calificaciones.csv")

# Describir la estructura del dataset
str(calificaciones)

# Calcular el vector de medias
medias <- colMeans(calificaciones, na.rm = TRUE)
print(medias)

# Calcular el vector de desviaciones típicas
desv_tip <- apply(calificaciones, 2, sd, na.rm = TRUE)
print(desv_tip)

# Identificar las actividades con mayor y menor variabilidad
actividad_mayor_variabilidad <- names(which.max(desv_tip))
actividad_menor_variabilidad <- names(which.min(desv_tip))

cat("Actividad con mayor variabilidad:", actividad_mayor_variabilidad, "\n")
cat("Actividad con menor variabilidad:", actividad_menor_variabilidad, "\n")
```

<div style="background-color:#5DADE2"><b>Apartado 2</b></div>

```r
# Calcular la matriz de correlación de Pearson
correlation_matrix <- cor(calificaciones, use = "complete.obs")

# Imprimir la matriz de correlación
correlation_matrix
```

<div style="background-color:#5DADE2"><b>Apartado 3</b></div>

```r
# Encontrar el par de variables con mayor correlación
max_correlation <- max(correlation_matrix[lower.tri(correlation_matrix)])
variables_max_corr <- which(correlation_matrix == max_correlation, arr.ind = TRUE)
variable1_max <- rownames(correlation_matrix)[variables_max_corr[1,1]]
variable2_max <- colnames(correlation_matrix)[variables_max_corr[1,2]]

cat("Par de variables con mayor correlación:", variable1_max, "y", variable2_max, "\n")

# Encontrar el par de variables con menor correlación (excluyendo la diagonal)
min_correlation <- min(correlation_matrix[lower.tri(correlation_matrix)])
variables_min_corr <- which(correlation_matrix == min_correlation, arr.ind = TRUE)
variable1_min <- rownames(correlation_matrix)[variables_min_corr[1,1]]
variable2_min <- colnames(correlation_matrix)[variables_min_corr[1,2]]


cat("Par de variables con menor correlación:", variable1_min, "y", variable2_min, "\n")
```
