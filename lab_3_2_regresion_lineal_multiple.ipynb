### Ejercicio 1

El dataframe `Boston` de la librería $\tt MASS$ contiene información de propiedades inmobiliarias de la ciudad de Boston. Cada propiedad está caracterizada por $13$ variables que indican las características de la misma, así como por la variable `medv` que representa una estimación del valor de la propiedad en miles de dólares. Considerando la variable dependiente `medv` estimar un modelo de regresión lineal múltiple para las siguientes variables independientes: `crim` que representa la tasa de crimen per capita de los distritos a los que pertenece cada una de las viviendas, la variable `dis` que representa la distancia ponderada a cinco centros de empleo de Boston y la variable `rm` que representa el número medio de habitaciones por vivienda. Realizar los siguientes apartados:

1. Exploración inicial de la base de datos
2. Análisis de la existencia de relación lineal entre las variables seleccionadas
3. Estimación de un modelo de regresión lineal múltiple entre las variables seleccionadas

<div style="background-color:#F7DC6F"><b>Resolución Ejercicio 01</b></div>

<div style="background-color:#5DADE2"><b>Apartado 1</b></div>

```r
# Cargar la librería MASS
library(MASS)

# Exploración inicial de la base de datos
# Ver las primeras filas del dataframe
head(Boston)

# Obtener información general del dataframe
str(Boston)

# Resumen estadístico de las variables
summary(Boston)

# Matriz de correlación de las variables
cor(Boston)

# Visualización de la relación entre las variables
pairs(Boston[, c("medv", "crim", "dis", "rm")])
```

<div style="background-color:#5DADE2"><b>Apartado 2</b></div>

```r
# Gráficos de dispersión para analizar la relación lineal entre las variables
plot(Boston$crim, Boston$medv, main = "Relación entre crim y medv", xlab = "crim", ylab = "medv")
plot(Boston$dis, Boston$medv, main = "Relación entre dis y medv", xlab = "dis", ylab = "medv")
plot(Boston$rm, Boston$medv, main = "Relación entre rm y medv", xlab = "rm", ylab = "medv")


# Cálculo de la correlación entre las variables
cor(Boston$crim, Boston$medv)
cor(Boston$dis, Boston$medv)
cor(Boston$rm, Boston$medv)
```

Interpretación de los resultados
Si la correlación entre dos variables es cercana a 1 o -1, indica una fuerte relación lineal positiva o negativa, respectivamente.
Si la correlación es cercana a 0, indica que no existe una relación lineal significativa entre las variables.

Se puede observar en los gráficos de dispersión y en la correlación que existe una relación lineal negativa entre crim y medv.
Existe una relación lineal positiva entre rm y medv, pero no tan fuerte como la relación entre crim y medv.
Finalmente existe una relación lineal negativa entre dis y medv, pero menos fuerte que la relación entre crim y medv.

### Ejercicio 2

Utilizando el dataset `Boston` encontrar de forma automática el mejor modelo de regresión lineal múltiple de la variable `medv` en función del resto de variables excepto `chas` y `rad`. Comparar sus métricas y realizar la diagnosis del modelo de regresión elegido. Es decir, tenéis que hacer los siguientes pasos:

1. Criterios de selección del modelo en base al ajuste de los datos
2. Proceso de automatización del proceso de selección del mejor modelo
3. Requisitos y validación del mejor modelo seleccionado en base al ajuste de los datos

<div style="background-color:#F7DC6F"><b>Resolución Ejercicio 02</b></div>

<div style="background-color:#5DADE2"><b>Apartado 1</b></div>

```r
# Cargar la librería MASS
library(MASS)

# Instalar y cargar la librería leaps para la selección de variables
if(!require(leaps)){
  install.packages("leaps")
  library(leaps)
}

# Crear una fórmula para el modelo de regresión con todas las variables excepto chas y rad
formula <- as.formula(paste("medv ~ . - chas - rad", sep = ""))

# Utilizar la función regsubsets para seleccionar el mejor modelo
modelo_seleccion <- regsubsets(formula, data = Boston, nbest = 1, nvmax = 11, method = "exhaustive")

# Resumen de los modelos seleccionados
summary(modelo_seleccion)

# Obtener el mejor modelo según el criterio de selección (bic)
best_model_bic <- which.min(summary(modelo_seleccion)$bic)
# Obtener el mejor modelo según el criterio de selección (adjusted R-squared)
best_model_adjR <- which.max(summary(modelo_seleccion)$adjr2)

# Crear el modelo con las variables seleccionadas
variables_seleccionadas_bic <- coef(modelo_seleccion, best_model_bic)
variables_seleccionadas_adjR <- coef(modelo_seleccion, best_model_adjR)


variables_seleccionadas_bic <- names(variables_seleccionadas_bic)[variables_seleccionadas_bic!=0]
variables_seleccionadas_adjR <- names(variables_seleccionadas_adjR)[variables_seleccionadas_adjR!=0] #Completing the line to extract the variable names for the adjusted R-squared model.
```

<div style="background-color:#5DADE2"><b>Apartado 2</b></div>

```r
# Cargar la librería MASS
library(MASS)

# Instalar y cargar la librería leaps para la selección de variables
if(!require(leaps)){
  install.packages("leaps")
  library(leaps)
}

# Crear una fórmula para el modelo de regresión con todas las variables excepto chas y rad
formula <- as.formula(paste("medv ~ . - chas - rad", sep = ""))

# Utilizar la función regsubsets para seleccionar el mejor modelo
modelo_seleccion <- regsubsets(formula, data = Boston, nbest = 1, nvmax = 11, method = "exhaustive")

# Resumen de los modelos seleccionados
summary(modelo_seleccion)

# Obtener el mejor modelo según el criterio de selección (bic)
best_model_bic <- which.min(summary(modelo_seleccion)$bic)
# Obtener el mejor modelo según el criterio de selección (adjusted R-squared)
best_model_adjR <- which.max(summary(modelo_seleccion)$adjr2)

# Crear el modelo con las variables seleccionadas
variables_seleccionadas_bic <- coef(modelo_seleccion, best_model_bic)
variables_seleccionadas_adjR <- coef(modelo_seleccion, best_model_adjR)


variables_seleccionadas_bic <- names(variables_seleccionadas_bic)[variables_seleccionadas_bic!=0]
variables_seleccionadas_adjR <- names(variables_seleccionadas_adjR)[variables_seleccionadas_adjR!=0] #Completing the line to extract the variable names for the adjusted R-squared model.
```

<div style="background-color:#5DADE2"><b>Apartado 3</b></div>

```r
# Cargar librerías necesarias
library(MASS)  # Para el dataset Boston

# Verificar si las variables seleccionadas existen y no están vacías
if (!exists("variables_seleccionadas_bic") || length(variables_seleccionadas_bic) == 0) {
    stop("Error: 'variables_seleccionadas_bic' no está definido o está vacío.")
}
if (!exists("variables_seleccionadas_adjR") || length(variables_seleccionadas_adjR) == 0) {
    stop("Error: 'variables_seleccionadas_adjR' no está definido o está vacío.")
}

# Eliminar posibles referencias a "Intercept"
variables_seleccionadas_bic <- setdiff(variables_seleccionadas_bic, "(Intercept)")
variables_seleccionadas_adjR <- setdiff(variables_seleccionadas_adjR, "(Intercept)")

# Construir los modelos con las variables seleccionadas
formula_bic <- as.formula(paste("medv ~", paste(variables_seleccionadas_bic, collapse = " + ")))
modelo_bic <- lm(formula_bic, data = Boston)

formula_adjR <- as.formula(paste("medv ~", paste(variables_seleccionadas_adjR, collapse = " + ")))
modelo_adjR <- lm(formula_adjR, data = Boston)

# Mostrar resúmenes de los modelos
summary(modelo_bic)
summary(modelo_adjR)

# Comparar las métricas de los modelos
print(paste("R-cuadrado ajustado (BIC):", summary(modelo_bic)$adj.r.squared))
print(paste("AIC (BIC):", AIC(modelo_bic)))
print(paste("BIC (BIC):", BIC(modelo_bic)))
print(paste("R-cuadrado ajustado (adjR):", summary(modelo_adjR)$adj.r.squared))
print(paste("AIC (adjR):", AIC(modelo_adjR)))
print(paste("BIC (adjR):", BIC(modelo_adjR)))

# Diagnóstico del modelo seleccionado (modelo basado en BIC)
par(mfrow = c(2,2))  # Configurar ventana gráfica para múltiples gráficos
plot(modelo_bic)  # Generar gráficos de diagnóstico

# Validación del modelo
set.seed(123)
indices <- sample(1:nrow(Boston), size = floor(0.7 * nrow(Boston)))
datos_entrenamiento <- Boston[indices, ]
datos_prueba <- Boston[-indices, ]

# Entrenar el modelo con los datos de entrenamiento
modelo_bic_entrenamiento <- lm(formula_bic, data = datos_entrenamiento)

# Predecir valores en los datos de prueba
predicciones <- predict(modelo_bic_entrenamiento, newdata = datos_prueba)

# Evaluar el error del modelo
error_rmse <- sqrt(mean((predicciones - datos_prueba$medv)^2))
print(paste("Error RMSE en datos de prueba:", error_rmse))
```
