### Ejercicio 3

El dataframe `Boston` de la librería $\tt MASS$ contiene información de propiedades inmobiliarias de la ciudad de Boston. Cada propiedad está caracterizada por $13$ variables que indican las características de la misma, así como por la variable `medv` que representa una estimación del valor de la propiedad en miles de dólares. Considerando la variable dependiente `medv` en función de la variable independiente `crim` que representa la tasa de crimen per capita de los distritos a los que petenece cada una de las viviendas. Se os pide responder a los siguientes apartados.

1. Exploración inicial de la base de datos.
2. Estimar dos o tres modelos de regresión polinómicos utilizando diferentes grados e interpretar sus resultados.
3. Compararlos gráficamente con el modelo de regresión lineal simple y calcular las métricas de bondad de ajuste para todos.
4. Estimar dos o tres modelos de regresión step functions utilizando diferentes puntos de corte e interpretar sus resultados.

<div style="background-color:#F7DC6F"><b>Resolución Ejercicio 03</b></div>

<div style="background-color:#5DADE2"><b>Apartado 1</b></div>

```r
# Carga de la librería MASS y el dataset Boston
library(MASS)
data(Boston)

# Exploración inicial de la base de datos
# Resumen estadístico de las variables
summary(Boston)

# Estructura del dataset
str(Boston)

# Dimensión del dataset
dim(Boston)

# Primeras filas del dataset
head(Boston)

# Visualización de la relación entre medv y crim
plot(Boston$crim, Boston$medv,
     xlab = "Tasa de crimen per cápita (crim)",
     ylab = "Valor medio de las viviendas (medv)",
     main = "Relación entre tasa de crimen y valor de la vivienda")


# Matriz de correlación
cor(Boston)

# Histograma de la variable medv
hist(Boston$medv, main = "Histograma de medv", xlab = "medv")

# Diagrama de dispersión de medv vs crim
plot(Boston$crim, Boston$medv, main="Dispersion de medv vs crim", xlab="crim", ylab="medv")

# Calculo de la correlación entre crim y medv
cor(Boston$crim, Boston$medv)
```

<div style="background-color:#5DADE2"><b>Apartado 2</b></div>

```r
# Modelo polinómico de grado 2
modelo_polinomico2 <- lm(Ozone ~ poly(Temp, 2), data = airquality)
summary(modelo_polinomico2)

# Modelo polinómico de grado 3
modelo_polinomico3 <- lm(Ozone ~ poly(Temp, 3), data = airquality)
summary(modelo_polinomico3)

# Modelo polinómico de grado 4 (ejemplo adicional)
modelo_polinomico4 <- lm(Ozone ~ poly(Temp, 4), data = airquality)
summary(modelo_polinomico4)
```

## Interpretación de los resultados
Comparar R-cuadrado, AIC, BIC para determinar el mejor modelo.
Analizar la significancia de los coeficientes.
Un mayor R-cuadrado indica mejor ajuste, pero hay que tener en cuenta la complejidad del modelo (grado del polinomio).
AIC y BIC penalizan la complejidad, por lo que modelos con menor AIC/BIC suelen ser preferibles.

Ejemplo de interpretación para el modelo de grado 2:
El coeficiente de determinación (R-squared) indica la proporción de la varianza en Ozone explicada por el modelo.
Los p-valores de los coeficientes indican si son estadísticamente significativos.
Si el p-valor es menor que el nivel de significancia (típicamente 0.05), se rechaza la hipótesis nula y se considera que el coeficiente es significativo.

<div style="background-color:#5DADE2"><b>Apartado 3</b></div>

```r
# Modelo de regresión lineal simple
modelo_lineal <- lm(Ozone ~ Temp, data = airquality)
summary(modelo_lineal)

# Gráfico de comparación
library(ggplot2)

ggplot(airquality, aes(x = Temp, y = Ozone)) +
  geom_point(alpha = 0.5) +  # Puntos de datos
  geom_smooth(method = "lm", se = FALSE, color = "blue", linetype = "dashed", formula = y ~ x) + # Regresión lineal simple
  geom_smooth(method = "lm", se = FALSE, color = "red", formula = y ~ poly(x, 2)) + # Regresión polinómica de grado 2
  geom_smooth(method = "lm", se = FALSE, color = "green", formula = y ~ poly(x, 3)) + # Regresión polinómica de grado 3
  geom_smooth(method = "lm", se = FALSE, color = "purple", formula = y ~ poly(x, 4)) + # Regresión polinómica de grado 4
  labs(title = "Comparación de modelos de regresión", x = "Temperatura", y = "Ozono") +
  theme_bw()


# Métricas de bondad de ajuste
# R-cuadrado
r_cuadrado_lineal <- summary(modelo_lineal)$r.squared
r_cuadrado_poli2 <- summary(modelo_polinomico2)$r.squared
r_cuadrado_poli3 <- summary(modelo_polinomico3)$r.squared
r_cuadrado_poli4 <- summary(modelo_polinomico4)$r.squared

# AIC
aic_lineal <- AIC(modelo_lineal)
aic_poli2 <- AIC(modelo_polinomico2)
aic_poli3 <- AIC(modelo_polinomico3)
aic_poli4 <- AIC(modelo_polinomico4)

# BIC
bic_lineal <- BIC(modelo_lineal)
bic_poli2 <- BIC(modelo_polinomico2)
bic_poli3 <- BIC(modelo_polinomico3)
bic_poli4 <- BIC(modelo_polinomico4)


# Crear una tabla para comparar las métricas
data.frame(
  Modelo = c("Lineal", "Polinómico Grado 2", "Polinómico Grado 3", "Polinómico Grado 4"),
  R_cuadrado = c(r_cuadrado_lineal, r_cuadrado_poli2, r_cuadrado_poli3, r_cuadrado_poli4),
  AIC = c(aic_lineal, aic_poli2, aic_poli3, aic_poli4),
  BIC = c(bic_lineal, bic_poli2, bic_poli3, bic_poli4)
)
```

<div style="background-color:#5DADE2"><b>Apartado 4</b></div>

```r
# Estimar dos o tres modelos de regresión step functions utilizando diferentes puntos de corte
library(splines)

# Modelo 1: Punto de corte en 70
modelo_step1 <- lm(Ozone ~ bs(Temp, knots = 70, degree = 1), data = airquality)
summary(modelo_step1)

# Modelo 2: Puntos de corte en 60 y 80
modelo_step2 <- lm(Ozone ~ bs(Temp, knots = c(60, 80), degree = 1), data = airquality)
summary(modelo_step2)

# Modelo 3: Puntos de corte en 50, 70 y 90
modelo_step3 <- lm(Ozone ~ bs(Temp, knots = c(50, 70, 90), degree = 1), data = airquality)
summary(modelo_step3)

# Visualización de los modelos
ggplot(airquality, aes(x = Temp, y = Ozone)) +
  geom_point(alpha = 0.5) +
  geom_smooth(method = "lm", formula = y ~ bs(x, knots = 70, degree = 1), se = FALSE, color = "blue") +
  geom_smooth(method = "lm", formula = y ~ bs(x, knots = c(60, 80), degree = 1), se = FALSE, color = "red") +
  geom_smooth(method = "lm", formula = y ~ bs(x, knots = c(50, 70, 90), degree = 1), se = FALSE, color = "green") +
  labs(title = "Modelos de regresión step functions", x = "Temperatura", y = "Ozono")


# Métricas de bondad de ajuste para los modelos step functions
# R-cuadrado
r_cuadrado_step1 <- summary(modelo_step1)$r.squared
r_cuadrado_step2 <- summary(modelo_step2)$r.squared
r_cuadrado_step3 <- summary(modelo_step3)$r.squared # Corrected this line
```

<div style="background-color:#5DADE2"><b>Apartado 3</b></div>

```r
# Graphical comparison
library(ggplot2)

ggplot(airquality, aes(x = Temp, y = Ozone)) +
  geom_point(alpha = 0.5) +  # Data points
  geom_smooth(method = "lm", se = FALSE, color = "blue", linetype = "dashed", formula = y ~ x) + # Simple linear regression
  geom_smooth(method = "lm", se = FALSE, color = "red", formula = y ~ poly(x, 2)) + # Polynomial regression degree 2
  geom_smooth(method = "lm", se = FALSE, color = "green", formula = y ~ poly(x, 3)) + # Polynomial regression degree 3
  geom_smooth(method = "lm", se = FALSE, color = "purple", formula = y ~ poly(x, 4)) + # Polynomial regression degree 4
  labs(title = "Comparison of Regression Models", x = "Temperature", y = "Ozone") +
  theme_bw()


# Goodness-of-fit metrics
# R-squared
r_cuadrado_lineal <- summary(modelo_lineal)$r.squared
r_cuadrado_poli2 <- summary(modelo_polinomico2)$r.squared
r_cuadrado_poli3 <- summary(modelo_polinomico3)$r.squared
r_cuadrado_poli4 <- summary(modelo_polinomico4)$r.squared

# AIC
aic_lineal <- AIC(modelo_lineal)
aic_poli2 <- AIC(modelo_polinomico2)
aic_poli3 <- AIC(modelo_polinomico3)
aic_poli4 <- AIC(modelo_polinomico4)

# BIC
bic_lineal <- BIC(modelo_lineal)
bic_poli2 <- BIC(modelo_polinomico2)
bic_poli3 <- BIC(modelo_polinomico3)
bic_poli4 <- BIC(modelo_polinomico4)

# Create a table to compare the metrics
data.frame(
  Model = c("Linear", "Polynomial Degree 2", "Polynomial Degree 3", "Polynomial Degree 4"), #Closed the string and added the rest of the model names
  R_cuadrado = c(r_cuadrado_lineal, r_cuadrado_poli2, r_cuadrado_poli3, r_cuadrado_poli4),
  AIC = c(aic_lineal, aic_poli2, aic_poli3, aic_poli4),
  BIC = c(bic_lineal, bic_poli2, bic_poli3, bic_poli4)
)
```

<div style="background-color:#5DADE2"><b>Apartado 4</b></div>

```r
# Create a data frame to store the results of all models
model_comparison <- data.frame(
  Model = c("Linear", "Polynomial Degree 2", "Polynomial Degree 3", "Polynomial Degree 4",
            "Step Function 1", "Step Function 2", "Step Function 3",
            "Spline 1", "Spline 2", "Spline 3"),
  R_squared = c(r_cuadrado_lineal, r_cuadrado_poli2, r_cuadrado_poli3, r_cuadrado_poli4,
                r_cuadrado_step1, r_cuadrado_step2, r_cuadrado_step3,
                summary(modelo_spline1)$r.squared, summary(modelo_spline2)$r.squared, summary(modelo_spline3)$r.squared),
  AIC = c(aic_lineal, aic_poli2, aic_poli3, aic_poli4,
          AIC(modelo_step1), AIC(modelo_step2), AIC(modelo_step3),
          AIC(modelo_spline1), AIC(modelo_spline2), AIC(modelo_spline3)),
  BIC = c(bic_lineal, bic_poli2, bic_poli3, bic_poli4,
          BIC(modelo_step1), BIC(modelo_step2), BIC(modelo_step3),
          BIC(modelo_spline1), BIC(modelo_spline2), BIC(modelo_spline3))
)


# Print the model comparison table
print(model_comparison)

# Determine the best model based on AIC and BIC
best_model_aic <- model_comparison[which.min(model_comparison$AIC), "Model"]
best_model_bic <- model_comparison[which.min(model_comparison$BIC), "Model"]

# Print the best models
cat("Best model based on AIC:", best_model_aic, "\n")
cat("Best model based on BIC:", best_model_bic, "\n")
```
