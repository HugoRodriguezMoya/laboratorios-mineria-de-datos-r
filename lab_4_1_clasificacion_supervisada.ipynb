## Clasificación Binaria

Definición: Solo hay dos clases posibles (sí/no, 1/0, positivo/negativo…).

### Ejercicio 4

Utilizando el conjunto de datos `titanic_train` de la libería $\tt titanic$ estamos interesados en explicar a través de un modelo de regresión logística la probabilidad que una persona que viajaba a bordo del Titanic sobreviviera ($1$) o no ($0$) en función de la edad que tenían cada uno de los $891$ pasajeros de los $2224$ que embarcaron en el transatlántico británico RMS Titanic. La variable dependiente binaria sería `Survived` y la variable independiente `Age`. Realizar los siguientes apartados:

1. Exploración inicial de la base de datos
2. Análisis de la existencia de relación lineal entre las variables seleccionadas
3. Estimación de un modelo de regresión logístico simple entre las variables seleccionadas
4. Interpretación de las salidas del modelo de regresión logístico simple estimado

<div style="background-color:#F7DC6F"><b>Resolución Ejercicio 04</b></div>

<div style="background-color:#5DADE2"><b>Apartado 1</b></div>

```r
if(!require(titanic)){install.packages("titanic")}
library(titanic)
if(!require(ggplot2)){install.packages("ggplot2")}
library(ggplot2)

data("titanic_train")

head(titanic_train)
str(titanic_train)
summary(titanic_train)

sum(is.na(titanic_train$Age))

titanic_train <- titanic_train[!is.na(titanic_train$Age), ]


plot(titanic_train$Age, titanic_train$Survived,
     xlab = "Age", ylab = "Survived",
     main = "Scatter Plot of Age vs. Survived")


model <- glm(Survived ~ Age, data = titanic_train, family = "binomial")

summary(model)

exp(coef(model)[2])

titanic_train$pred_prob <- predict(model, type = "response")

ggplot(titanic_train, aes(x = Age, y = pred_prob)) +
  geom_point(alpha = 0.5) +
  geom_smooth(method = "glm", method.args = list(family = "binomial"), se = FALSE, color = "red") +
  labs(title = "Predicted Probability of Survival vs. Age", x = "Age", y = "Predicted Probability")
```

Carga de datos y limpieza:
Se carga el conjunto de datos y se eliminan las filas con valores faltantes en la variable Age para poder realizar el análisis.

Visualización inicial:
Se genera un gráfico de dispersión para observar la relación entre la edad y la variable binaria Survived.

Modelo logístico:
Se ajusta un modelo de regresión logística simple (glm) con Age como variable predictora de Survived.

Interpretación del modelo:
Se observa el resumen del modelo y se calcula el efecto de la edad sobre la probabilidad de supervivencia. El valor de exp(coef) indica el cambio en las probabilidades por cada año adicional de edad.

Visualización de la predicción:
Se calcula la probabilidad de supervivencia para cada observación y se grafica con ggplot2, mostrando cómo varía la probabilidad de sobrevivir según la edad, junto con una curva ajustada en rojo que representa el modelo logístico.

<div style="background-color:#5DADE2"><b>Apartado 3</b></div>

```r
model <- glm(Survived ~ Age, data = titanic_train, family = "binomial")

summary(model)
```

Se ajusta un modelo de regresión logística con Survived como variable dependiente y Age como variable independiente.
La función glm(..., family = "binomial") se utiliza porque la variable objetivo (Survived) es binaria (0 = no, 1 = sí).

El resumen del modelo (summary(model)) muestra los coeficientes, errores estándar y significancia estadística.
Un coeficiente negativo para Age indicaría que a mayor edad, menor probabilidad de supervivencia, y viceversa.

<div style="background-color:#5DADE2"><b>Apartado 4</b></div>

Interpretación de las salidas del modelo

La salida de summary(model) proporciona información clave sobre el modelo de regresión logística estimado.

1. Coeficientes:
  - La columna "Estimate" muestra los coeficientes estimados para la intersección (intercept) y la variable predictora (Age).
  - El coeficiente de Age (-0.028 en este ejemplo, aunque puede variar con las ejecuciones) indica la relación entre la edad y la probabilidad de supervivencia. Un valor negativo sugiere que a medida que aumenta la edad, la probabilidad de supervivencia disminuye (en este caso, por cada año adicional de edad, la probabilidad de supervivencia se multiplica por exp(-0.028)).
  - Es importante considerar el p-valor (columna "Pr(>|z|)") asociado a cada coeficiente para determinar su significancia estadística. Si el p-valor es menor que el nivel de significancia alfa (usualmente 0.05), se rechaza la hipótesis nula de que el coeficiente sea cero.

2. Estadísticos de bondad de ajuste:
  - Deviance:  Mide la discrepancia entre el modelo y los datos.
  - AIC (Akaike Information Criterion): Es una medida de ajuste que penaliza modelos complejos. Un menor AIC generalmente indica un modelo mejor.
  - Null deviance: Devianza del modelo nulo (modelo que solo incluye la intersección). La diferencia entre la devianza nula y la devianza del modelo indica la mejora en el ajuste que proporciona la variable predictora.
