<div style="background-color:#5DADE2"><b>Apartado 1</b></div>

El proceso de entrenamiento en los modelos de clasificación supervisada consiste en enseñar al modelo a reconocer patrones y relaciones en los datos. Para ello, se le proporciona un conjunto de ejemplos que ya tienen una etiqueta o clase conocida (conjunto de entrenamiento). A partir de esos ejemplos, el modelo ajusta sus parámetros internos para que sea capaz de predecir correctamente la clase de nuevos datos no vistos.

Es decir, el modelo "aprende" cómo se relacionan las variables (también llamadas características o features) con las clases, con el objetivo de generalizar ese conocimiento y poder aplicarlo a futuros casos.

Datos etiquetados: El entrenamiento se realiza con un conjunto de datos donde se conoce la clase o categoría correcta de cada ejemplo. Esto permite al modelo aprender comparando sus predicciones con las respuestas reales.

Algoritmo de aprendizaje: Se utiliza un algoritmo específico (como k-NN, árbol de decisión, regresión logística, etc.) que define cómo se actualizan los parámetros del modelo durante el aprendizaje.

Error o función de pérdida: El modelo mide cuánto se equivoca en cada predicción y utiliza ese error para ajustar su comportamiento. Cuanto menor sea el error, mejor ha aprendido el modelo.

Iteraciones o ajustes: En muchos casos, el modelo se entrena durante varias iteraciones, corrigiendo poco a poco sus errores hasta alcanzar un rendimiento aceptable.

Evaluación del rendimiento: Una parte importante del proceso es comprobar cómo de bien está funcionando el modelo. Esto se hace normalmente separando los datos en conjuntos de entrenamiento y prueba, para ver si el modelo generaliza bien a datos nuevos.

Evitar el sobreajuste (overfitting) y el subajuste (underfitting): El objetivo no es memorizar los datos de entrenamiento, sino aprender patrones generales. El sobreajuste ocurre cuando el modelo aprende demasiado bien el entrenamiento y falla en datos nuevos. El subajuste ocurre cuando el modelo no es lo suficientemente complejo para aprender bien.

<div style="background-color:#5DADE2"><b>Apartado 2</b></div>

1. k-Nearest Neighbors (k-NN)
Es un modelo muy intuitivo que clasifica un nuevo dato observando qué clase tienen sus k vecinos más cercanos (según una medida de distancia, como la euclídea). La clase más frecuente entre esos vecinos se asigna al nuevo dato.
2. Árboles de Decisión (Decision Trees)
Este modelo divide los datos en ramas siguiendo reglas de decisión (como "¿la edad > 30?") hasta llegar a una predicción final en forma de hoja. Es como un diagrama de decisiones.

3. Regresión Logística
Aunque su nombre contiene "regresión", es un modelo de clasificación. Estima la probabilidad de que un ejemplo pertenezca a una clase utilizando una función logística (sigmoide).

4. Naive Bayes
Se basa en el Teorema de Bayes, asumiendo que todas las variables son independientes entre sí (lo cual rara vez se cumple al 100%). Aun así, funciona sorprendentemente bien en muchos casos, especialmente en clasificación de texto.

5. Máquinas de Vectores de Soporte (SVM)
Busca encontrar el hiperplano que mejor separa las clases en el espacio de características. Puede usar kernels para separar datos que no son linealmente separables.

6. Redes Neuronales Artificiales (ANN)
Inspiradas en el cerebro humano, se basan en capas de nodos (neuronas) que transforman la información de entrada hasta producir una salida.

<div style="background-color:#5DADE2"><b>Apartado 3</b></div>

El sobreajuste (overfitting) y el infra-ajuste (underfitting) son dos problemas comunes al entrenar modelos de clasificación supervisada.

El sobreajuste ocurre cuando el modelo aprende demasiado bien los datos de entrenamiento, incluyendo detalles irrelevantes o ruido. Esto hace que el modelo funcione muy bien con los datos con los que ha sido entrenado, pero tenga un rendimiento muy bajo cuando se le presentan nuevos datos. Suele ser causado por un modelo demasiado complejo, como un árbol de decisión muy profundo o una red neuronal con muchas capas. Para evitarlo, se pueden usar técnicas como la regularización, la poda de árboles, la validación cruzada o incluso añadir más datos al entrenamiento.

Por otro lado, el infra-ajuste sucede cuando el modelo no es capaz de aprender ni siquiera los patrones básicos de los datos. En este caso, el rendimiento es malo tanto en los datos de entrenamiento como en los de prueba. Este problema suele deberse a que el modelo es demasiado simple para la complejidad del problema, como por ejemplo usar una regresión logística para un conjunto de datos con relaciones no lineales. Para solucionarlo, se puede aumentar la complejidad del modelo, mejorar la selección de variables o cambiar de algoritmo.
